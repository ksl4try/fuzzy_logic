```python
import numpy as np
%pip install scikit-fuzzy
import skfuzzy as fuzzy
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Definisikan atau atur koordinat untuk tipa variabel fuzzy
x_usia = np.arange(18, 58, 1)
x_suhubadan = np.arange(27.5, 45.0, 1)
x_kondnapas = np.arange(14, 24, 1)
x_jmlantbd = np.arange(0.1, 8.0, 1)
x_status = np.arange(0, 100, 1)

usia_muda = fuzzy.trapmf(x_usia, [18, 18, 30, 35])
usia_pbaya = fuzzy.trapmf(x_usia, [30, 36, 45, 50])
usia_tua = fuzzy.trapmf(x_usia, [45, 50, 58, 58])

suhu_hipo = fuzzy.trapmf(x_suhubadan, [27.5, 27.5, 34.5, 35.5])
suhu_normal = fuzzy.trapmf(x_suhubadan, [34.5, 35.5, 36.5, 37.5])
suhu_hiper = fuzzy.trapmf(x_suhubadan, [36.5, 37.5, 45.0, 45.0])

kondnapas_normal = fuzzy.trapmf(x_kondnapas, [14, 14, 16, 17])
kondnapas_moderat = fuzzy.trapmf(x_kondnapas, [16, 18, 19, 20])
kondnapas_abnormal = fuzzy.trapmf(x_kondnapas, [19, 20, 24, 24])

jmlantbd_normal = fuzzy.trapmf(x_jmlantbd, [0.1, 0.1, 1.5, 2.0])
jmlantbd_moderat = fuzzy.trapmf(x_jmlantbd, [1.5, 2.0, 3.8, 4.0])
jmlantbd_abnormal = fuzzy.trapmf(x_jmlantbd, [3.8, 4.0, 8.0, 8.0])

status_negatif = fuzzy.trapmf(x_status, [0, 0, 49, 50])
status_positif = fuzzy.trapmf(x_status, [50, 51, 100, 100])
```

```python
fig, (ax0, ax1, ax2, ax3, ax4) = plt.subplots(nrows=5, figsize=(8, 10))

ax0.plot(x_usia, usia_muda, 'b', linewidth=1.5, label='Muda')
ax0.plot(x_usia, usia_pbaya, 'g', linewidth=1.5, label='Paruhbaya')
ax0.plot(x_usia, usia_tua, 'r', linewidth=1.5, label='Tua')
ax0.set_title('Indeks Usia Pasien')
ax0.legend

ax1.plot(x_suhubadan, suhu_hipo, 'b', linewidth=1.5, label='Hipotermia')
ax1.plot(x_suhubadan, suhu_normal, 'g', linewidth=1.5, label='Normal')
ax1.plot(x_suhubadan, suhu_hiper, 'r', linewidth=1.5, label='Hipertermia')
ax1.set_title('Indeks Suhu Badan')
ax1.legend

ax2.plot(x_kondnapas, kondnapas_normal, 'b', linewidth=1.5, label='Normal')
ax2.plot(x_kondnapas, kondnapas_moderat, 'g', linewidth=1.5, label='Moderat')
ax2.plot(x_kondnapas, kondnapas_abnormal, 'r', linewidth=1.5, label='Abnormal')
ax2.set_title('Indeks Kondisi Pernapasan')
ax2.legend

ax3.plot(x_jmlantbd, jmlantbd_normal, 'b', linewidth=1.5, label='Normal')
ax3.plot(x_jmlantbd, jmlantbd_moderat, 'g', linewidth=1.5, label='Moderat')
ax3.plot(x_jmlantbd, jmlantbd_abnormal, 'r', linewidth=1.5, label='Abnormal')
ax3.set_title('Indeks Jumlah Antibodi')
ax3.legend

ax4.plot(x_status, status_negatif, 'b', linewidth=1.5, label='Negatif')
ax4.plot(x_status, status_positif, 'g', linewidth=1.5, label='Positif')
ax4.set_title('Indeks Status Pasien')
ax4.legend

for ax in (ax0, ax1, ax2, ax3, ax4):
  ax.spines['top'].set_visible(False)
  ax.spines['right'].set_visible(False)
  ax.get_xaxis().tick_bottom()
  ax.get_yaxis().tick_left()
  plt.tight_layout()
```

```python
# Import library
import numpy as np
import matplotlib.pyplot as plt

# Domain umur (0 - 80 tahun)
x = np.linspace(0, 80, 500)

# Fungsi keanggotaan fuzzy untuk umur
def muda(x):
    if x <= 25:
        return 1
    elif 25 < x <= 35:
        return (35 - x) / (35 - 25)
    else:
        return 0

def dewasa(x):
    if 35 < x <= 45:
        return (x - 35) / (45 - 35)
    elif 45 < x <= 55:
        return (55 - x) / (55 - 45)
    else:
        return 0

def tua(x):
    if x <= 55:
        return 0
    elif 55 < x <= 65:
        return (x - 55) / (65 - 55)
    else:
        return 1

# Konversi ke bentuk array
muda_values = np.array([muda(val) for val in x])
dewasa_values = np.array([dewasa(val) for val in x])
tua_values = np.array([tua(val) for val in x])

# Plot grafik
plt.figure(figsize=(10,6))
plt.plot(x, muda_values, label='Muda', color='blue', linewidth=2)
plt.plot(x, dewasa_values, label='Dewasa', color='green', linewidth=2)
plt.plot(x, tua_values, label='Tua', color='red', linewidth=2)

# Garis putus-putus (batas penting)
plt.axvline(x=25, color='blue', linestyle='--', linewidth=1)
plt.axvline(x=45, color='green', linestyle='--', linewidth=1)
plt.axvline(x=65, color='red', linestyle='--', linewidth=1)

# Label titik batas
plt.text(25, 0.05, '25 (Muda)', rotation=90, color='blue', va='bottom', ha='right')
plt.text(45, 0.05, '45 (Dewasa)', rotation=90, color='green', va='bottom', ha='right')
plt.text(65, 0.05, '65 (Tua)', rotation=90, color='red', va='bottom', ha='right')

# Tambahan tampilan
plt.title('Indeks Usia')
plt.xlabel('Umur (tahun)')
plt.ylabel('Derajat Keanggotaan μ[x]')
plt.grid(True, linestyle=':', linewidth=0.7)
plt.legend()
plt.ylim(-0.05, 1.05)
plt.xlim(0, 80)
plt.show()
```
